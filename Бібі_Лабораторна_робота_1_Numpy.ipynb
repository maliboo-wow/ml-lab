### Завдання 28
Для матриці:

```text

     [2, 6]]
```

обчисліть:
- визначник;
- обернену матрицю.

Перевірте результат, помноживши початкову матрицю на обернену.

````python
# Код студента
A = np.array([[4, 7], [2, 6]])

inv_A = np.linalg.inv(A)

print("визначник:", np.linalg.det(A))
print("обернена:\n", inv_A)
print("перевірка A @ inv:\n", (A @ inv_A).round(3))   # має бути одинична
````

### Завдання 30
Створіть одиничну матрицю розміром `4 × 4`.

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

mean = np.mean(A, axis=0)
std = np.std(A, axis=0)

print((A - mean) / std)
````

### Завдання 39
Створіть матрицю `4 × 4` із довільних цілих чисел.

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

Спробуйте виконати завдання без циклів.

````python
# Код студента
np.random.seed(1)
A = np.random.randint(1, 10, (4, 4)).astype(float)

norms = np.linalg.norm(A, axis=1)  # норма кожного рядка
normalized = A / norms.reshape(4, 1) # ділимо кожен рядок на його норму

print("Початкова матриця:")
print(A)
print("Норми рядків:", norms)
print("Нормалізована матриця:")
print(normalized)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента
A = np.array([[1, 2, 1],
              [2, 1, -1],
              [3, -1, 2]])
b = np.array([9, 8, 3])
x = np.linalg.solve(A, b)

print("розв'язок:", x)
print("перевірка A @ x:", (A @ x).round(3))
print("b:", b)
````
